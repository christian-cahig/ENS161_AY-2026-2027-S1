```python
import math as mt
import numpy as np
import numpy.linalg as npla
```

## Scenario 1

Observations.
1. $EF$ is a zero-force member.
2. $BF$ and $EF$ are collinear.
3. $CD$ and $BC$ are collinear.
4. The line of action of the 7-kip load, $CF$, and $AF$ are collinear.

From Observation 1,
$DE$ and $EA$ carry the same forces.

From Observations 2 and 4,
- $BF$ is also a zero-force member,
  and
- $CF$ and $AF$ carry the same forces.

From Observations 3 and 4,
- $CD$ and $BC$ carry the same forces,
  and
- $CF$ directly transmits the 7-kip laod.

Alas, we can method-of-joints our way through the truss.
Start at $D$,
and you will get the forces in $DE$ and $CD$,
which are then respectively transmitted by $EA$ and $BC$.
Balancing forces at $A$ and at $B$ will yield the reactions thereat.

```python
# Force balance at D
A = np.array([[3, -3], [4, 4]])
b = np.array([-4, -3]) * 2.45
# Tensile forces in DE and CD
x = npla.solve(A, b)
T_DE, T_CD = x[0], x[1]

T_AE = T_DE; T_BC = T_CD
T_AF = -7; T_BF = 0

# Force balance at A
Ax = -(T_AE * 3/5) - (T_AF * 4/5)
Ay = -(T_AE * 4/5) - (T_AF * 3/5)

# Force balance at B
Bx = 6 + (4.55 * 4/5) + (T_BC * 3/5)
By = (4.55 * 3/5) - (T_BC * 4/5) - (T_BF * 3/5)

print(f"a. AE = {T_AE * 1e3} lb")
print(f"b. AF = {T_AF * 1e3} lb")
print(f"c. BF = {T_BF * 1e3} lb")
print(f"d. BC = {T_BC * 1e3} lb")
print(f"e. rxn at A: <{Ax * 1e3}; {Ay * 1e3}> lb")
print(f"f. rxn at B: <{Bx * 1e3}; {By * 1e3}> lb")
```

## Scenario 2

Observations
1. $BG$ is a zero-force member.
2. $BC$ and $AB$ are collinear.

$BC$ and $AB$ will thus carry the same forces.

```python
R = 0.25; P = 50
_sqrt2 = mt.sqrt(2)

# Cable tension
T = P / 2

T_CD = -P * _sqrt2
T_FG = T
T_CG = -P * _sqrt2
T_BC = -P
T_EF = T * ((2 * _sqrt2) - 1)
T_GH = T * _sqrt2
T_AG = T * ((3 * _sqrt2) - 1)

Hx = -T
Hy = -T

Ax = T * (2 - _sqrt2) / 2
Ay = T * (6 + _sqrt2) / 2

print(f"a. cable: {T} kN")
print(f"b. CD = {T_CD} kN")
print(f"c. FG = {T_FG} kN")
print(f"d. CG = {T_CG} kN")
print(f"e. BC = {T_BC} kN")
print(f"f. EF = {T_EF} kN")
print(f"g. GH = {T_GH} kN")
print(f"h. AG = {T_AG} kN")
print(f"i. rxn at H: <{Hx}; {Hy}> kN")
print(f"j. rxn at A: <{Ax}; {Ay}> kN")
```

## Scenario 3

One may method-of-joints their way
- first at $E$
  (to solve for the forces in $EF$ and $DE$),
- then at $F$
  (to solve for the forces in $GF$ and $DF$),
  and
- finally at $D$
  (to solve for the forces in $DG$ and $CD$).

This approach, while effective, might be too "monocultural" for one who likes to mix methods up.
For the latter case, one may
- first pass a section through $GF$, $DF$, and $DE$
  (to solve for the forces in those members),
- then balance forces at $E$
  (to solve for the force in $EF$),
  and
- finally pass a section through $GF$, $DG$, and $CD$
  (to solve for the forces in $DG$ and $CD$).

```python
alpha = mt.radians(22.5)
alpha2 = 2 * alpha
salpha = mt.sin(alpha)
calpha = mt.cos(alpha)
calpha2 = mt.cos(alpha2)
theta = (mt.pi - alpha) / 2
stheta = mt.sin(theta)
phi = (mt.pi - alpha2) / 2
sphi = mt.sin(phi)
beta = mt.atan((2 * salpha) / ((2 * calpha) - 1))
sbeta = mt.sin(beta)

L = 1

T_DE = L / stheta
T_EF = -T_DE * mt.cos(theta)
T_GF = L * ((4 * calpha) - 3) / (2 - calpha)
T_DF = (L - (T_GF * sphi)) / sbeta
T_DG = ((((4 * calpha2) - 3) * L) - (T_GF * sphi)) / sbeta
T_CD = L * (3 - (2 * calpha2)) / stheta

print(f"a. DE = {T_DE}L")
print(f"b. EF = {T_EF}L")
print(f"c. GF = {T_GF}L")
print(f"d. DF = {T_DF}L")
print(f"e. DG = {T_DG}L")
print(f"f. CD = {T_CD}L")
```

## Scenario 4

This is a K-truss, but tapered.
First, pass a section through $CF$, $FG$, $GH$, and $EH$.
Balance moments about $H$ to solve for the force in $CF$.
Balance moments about $F$ to solve for the force in $EH$.
Then, pass a section through $CF$, $CG$, $EG$, and $EH$.
Balance moments about $E$ to solve for the force in $CG$.
Balance moments about $C$ to solve for the force in $EG$.

Treating the entire truss as the rigid body,
first balance moments about $B$ to solve for the reaction at $A$,
and
balance forces to solve for the reaction at $B$.

```python
root10 = mt.sqrt(10)
root10_over_18 = root10 / 18
root13_over_36 = mt.sqrt(13) / 36

T_CF = ((30 * 9) - (50 * 3)) * root10_over_18
T_EH = (-(30 * 9) - (50 * 3)) * root10_over_18
T_CG = ((30 * 18) + (30 * 9) - (50 * 6) - (T_CF * 36 / root10)) * root13_over_36
T_EG = (-(30 * 9) - (30 * 18) - (50 * 6) - (T_EH * 36 / root10)) * root13_over_36

Ax = 0.
Ay = -65.
Bx = -90.
By = 115.

print(f"a. CF = {T_CF} k")
print(f"b. CG = {T_CG} k")
print(f"c. EG = {T_EG} k")
print(f"d. EH = {T_EH} k")
print(f"e. rxn at B: <{Bx}; {By}> k")
print(f"f. rxn at A: <{Ax}; {Ay}> k")
```

## Scenario 5

Like that in Scenario 4,
this is a tapered K-truss,
but with the K on its back with its legs up like a beetle.
First, pass a section through $DG$, $ED$, $EF$, and $FH$.
Balance moments about $D$ to solve for the force in $FH$.
Balance moments about $F$ to solve for the force in $DG$.
Then, pass a section through $DG$, $EG$, $EH$, and $FH$.
Balance moments about $G$ to solve for the force in $EH$.
Balance moments about $H$ to solve for the force in $EG$.

Treating the entire truss as the rigid body,
first balance moments about $C$ to solve for the reaction at $A$,
and
balance forces to solve for the reaction at $C$.

```python
root10 = mt.sqrt(10)
root10_over_18 = mt.sqrt(10) / 18
root10_over_36 = root10_over_18 / 2

T_DG = ((30 * 9) + (30 * 18) - (50 * 6)) * root10_over_36
T_FH = -((50 * 6) + (30 * 9) + (30 * 18)) * root10_over_36
T_EG = ((30 * 9) - (50 * 3) - (T_DG * 18 / root10)) * root10_over_18
T_EH = -((30 * 9) + (50 * 3) + (T_FH * 18 / root10)) * root10_over_18

Ax = 0.
Ay = -65.
Cx = -90.
Cy = 115.

print(f"a. DG = {T_DG} k")
print(f"b. EG = {T_EG} k")
print(f"c. EH = {T_EH} k")
print(f"d. FH = {T_FH} k")
print(f"e. rxn at A: <{Ax}; {Ay}> k")
print(f"f. rxn at C: <{Cx}; {Cy}> k")
```
