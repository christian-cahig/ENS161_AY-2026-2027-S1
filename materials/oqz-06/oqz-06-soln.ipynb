## Strategy

Analyze beam $CD$ on its own.
It is supported by the hinge at $C$ and the roller at $D$.
Balance moments about $C$
and you will have the roller reaction force $D_{y}$ (upward).
Balance forces
and you will have the hinge "reaction" forces $C_{x}$ (rightward) and $C_{y}$ (upward).

Now analyze the beam of concern $ABC$ on its own.
It is supported by the pin at $A$ and the roller at $B$.
Its "free" end $C$ is subject to forces $C_{x}$ (leftward) and $C_{y}$ (downward).
Balance moments about $A$
and you will have roller reaction force $B_{y}$ (upward).
Balance forces
and you will have the pin reaction forces $A_{x}$ (rightward) and $A_{y}$ (downward).

To derive the expressions for the internal reactions,
pass arbitrary sections within the intervals
$0 \leq x \leq l_{1}$
and
$l_{1} \leq x \leq l_{1} + l_{2} $.

## Answers

$$D_{y} \left(l_{3} + l_{4}\right) = \frac{4}{5} P l_{3}$$

$$C_{y} + D_{y} = \frac{4}{5} P$$

$$C_{x} = \frac{3}{5} P$$

$$B_{y} l_{1} = w \left(l_{1} + l_{2}\right) \left(\frac{l_{1} + l_{2}}{2}\right) + M + C_{y} \left(l_{1} + l_{2}\right)$$

$$A_{y} + C_{y} + w \left(l_{1} + l_{2}\right) = B_{y}$$

$$A_{x} = C_{x}$$

```python
l1, l2, l3, l4 = 3.0, 0.9, 1.0, 1.1
w, M, P = 1.0, 2.0, 12.
```

```python
Dy = (P * 0.8 * l3) / (l3 + l4)
Cy = (P * 0.8) - Dy
Cx = P * 0.6

l1l2 = l1 + l2; l1l2_ov2 = l1l2 / 2
By = ((w * l1l2 * l1l2_ov2) + M + (Cy * l1l2)) / l1
Ay = By - Cy - (w * l1l2)
Ax = Cx
```

For an arbitrary section passing somewhere within $0 \leq x < l_{1}$:

$$\eta \left(x\right) + A_{x} = 0$$

$$\nu \left(x\right) + A_{y} + wx = 0$$

$$\mu \left(x\right) + A_{y} x + wx \left(\frac{x}{2}\right) = 0$$

```python
eta = npp.Polynomial([-Ax])
print(f"η(x) = {eta} kip")

nu = npp.Polynomial([-Ay, -w])
print(f"ν(x) = {nu} kip")

mu = npp.Polynomial([0, -Ay, -w/2])
print(f"μ(x) = {mu} kip-ft")
```

For an arbitrary section passing somewhere within $l_{1} \leq x < l_{1} + l_{2}$:

$$\eta \left(x\right) + A_{x} = 0$$

$$\nu \left(x\right) + A_{y} + wx = B_{y}$$

$$\mu \left(x\right) + A_{y} x + wx \left(\frac{x}{2}\right) = M + B_{y} \left(x - l_{1}\right)$$

```python
eta = npp.Polynomial([-Ax])
print(f"η(x) = {eta} kip")

nu = npp.Polynomial([-Ay - By, -w])
print(f"ν(x) = {nu} kip")

mu = npp.Polynomial([M - (By * l1), By - Ay, -w/2])
print(f"μ(x) = {mu} kip-ft")
```
