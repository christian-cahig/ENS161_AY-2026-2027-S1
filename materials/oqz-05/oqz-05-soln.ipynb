## Scenario 1

First, consider beam $BCD$,
thus treating the hinge force as a support reaction.
Balance moments about $B$ to solve for the reaction at $C$.
Balance forces to solve for the reaction force at the hinge.

Then, consider beam $AB$,
now treating the hinge force as an applied force.
The direction of the hinge force here is opposite to that assumed in considering beam $BCD$.
Balance forces to solve for the reaction force at $A$.
Balance moments about $A$ to solve for the reaction moment at $A$.

```python
l1, l2, l3, w, P, M = 14, 15, 16, 2.3, 5, 30
wl1 = w * l1; wl2 = w * l2

Cx = 0
Cy = ((wl2 * l2 / 2) + (P * (l2 + l3)) + M) / l2

Bx = 0
By = P + wl2 - Cy

Ax = Bx
Ay = wl1 + By
Am = (wl1 * l1 / 2) + (By * l1)
```

```python
print(f"a. reaction force at A: <{Ax:.7f}; {Ay:.7f}> k")
print(f"b. reaction moment at A: <{0:.7f}; {0:.7f}; {Am:.7f}> k-ft")
print(f"c. hinge force at B: <{Bx:.7f}; {By:.7f}> k")
print(f"d. reaction at C: <{Cx:.7f}, {Cy:.7f}> k")
```

## Scenario 2

First, consider beam $CD$
thus treating the hinge force as a support reaction.
Balance moments about $C$ to solve for the reaction at $D$.
Balance forces to solve for the reaction force at the hinge.

Then, consider beam $ABC$,
now treating the hinge force as an applied force.
The direction of the hinge force here is opposite to that assumed in considering beam $CD$.
Balance moments about $A$ to solve for the reaction at $B$.
Balance forces to solve for the reaction at $A$.

```python
l1, l2, l3, l4, w, P = 30, 9, 10, 11, 1.0, 12
l1l2 = l1 + l2
wl1l2 = w * l1l2

Dx = 0
Dy = (P * l3) / (l3 + l4)

Cx = 0
Cy = P - Dy

Bx = 0
By = ((wl1l2 * l1l2 / 2) + (Cy * l1l2)) / l1

Ax = 0
Ay = wl1l2 + Cy - By
```

```python
print(f"a. reaction at A: <{Ax:.7f}; {Ay:.7f}> k")
print(f"b. reaction at B: <{Bx:.7f}; {By:.7f}> k")
print(f"c. hinge force at C: <{Cx:.7f}; {Cy:.7f}> k")
print(f"d. reaction at D: <{Dx:.7f}; {Dy:.7f}> k")
```

## Scenario 3

Consider the entire compound frame $ACDEB$.
Balance moments about $A$ to solve for the vertical component of the reaction at $B$.
Similarly, balance moments about $B$ to solve for the vertical component of the reaction at $A$.

Consider $ACD$ and treat the hinge force as support reaction.
Balance moments about $D$ to solve for the horizontal component of the reaction at $A$.
Balance forces to determine the reaction force at the hinge.

Consider $DEB$ and treat the hinge force as an applied force.
Remember to orient the hinge force opposite to the direction used when considering $ACD$.
Balance forces to solve for the horizontal component of the reaction at $B$.

```python
P, w, l1, l2, l3 = 30, 2, 13, 17, 21
l1l2 = l1 + l2
wl1 = w * l1; wl2 = w * l2
wl1l2 = wl1 + wl2

By = ((P * l3) + (wl1l2 * l1l2 / 2)) / l1l2
Ay = ((wl1l2 * l1l2 / 2) - (P * l3)) / l1l2

Ax = ((Ay * l1) - (wl1 * l1 / 2)) / l3

Dx = -P - Ax
Dy = wl1 - Ay

Bx = Dx
```

```python
print(f"a. reaction at A: <{Ax:.7f}; {Ay:.7f}> k")
print(f"b. reaction at B: <{Bx:.7f}; {By:.7f}> k")
print(f"c. hinge force at D: <{Dx:.7f}; {Dy:.7f}> k")
```

## Scenario 4

Consider $ABC$ and treat the hinge force as support reaction.
Balance moments about $C$ to solve for reaction at $A$.
Balance forces to solve for the reaction at the hinge.

Consider $CD$ and treat the hinge force as an applied force.
Be sure to orient the hinge force opposite to its direction when considering $ABC$.
Balance forces to solve for the reaction force at $D$.
Balance moments about $D$ to solve for the reaction moment at $D$.

```python
P, w, l1, l2, l3, l4 = 75, 15, 6, 7, 6, 9
wl3 = w * l3; wl4 = w * l4

Ax = 0
Ay = ((P * l2) + (wl3 * l3 / 2)) / (l1 + l2)

Cx = wl3
Cy = P - Ay

Dx = Cx + wl4
Dy = Cy
Dm = -(wl4 * l4 / 2) - (Cx * l4)
```

```python
print(f"a. reaction at A: <{Ax:.7f}; {Ay:.7f}> kN")
print(f"b. hinge force at C: <{Cx:.7f}; {Cy:.7f}> kN")
print(f"c. reaction force at D: <{Dx:.7f}; {Dy:.7f}> kN")
print(f"d. reaction moment at D: <{0:.7f}; {0:.7f}; {Dm:.7f}> kN-m")
```

## Scenario 5

Consider the blade and draw its FBD.
Let $V$ be the vertical contact force between the blade and the branch,
and $F$ be the force in $AB$.
Balance moments about $C$.
You should arrive at an equation in terms of $V$ and $F$.
Set $V = 317$ to solve for $F$.

Consider the lower lever
(*i.e.*, the one containing $B$ and $D$).
Balance moments about $D$.
You should arrive at an equation in terms of $P$ and $F$.
Use the value of $F$ from the previous step
and you will get the required value of $P$.

```python
d1, d2, d3, d4, d5, d6, d7 = 3.5, 1.6, 0.5, 0.55, 0.25, 0.65, 0.75
V = 317

d_AB = mt.sqrt((d6**2) + (d4**2))
F = (V * d2) / ((d3 * d6 / d_AB) + ((d6 + d7) * d4 / d_AB))

P = ((F * d7 * d4 / d_AB) - (F * d5 * d6 / d_AB)) / d1
```

```python
print(f"a. required P: {P:.7f} lb")
print(f"b. force in AB: {F:.7f} lb")
```
