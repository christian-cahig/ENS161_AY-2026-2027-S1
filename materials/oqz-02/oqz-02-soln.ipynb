```python
import math as mt

import numpy as np
import scipy.constants as spc
```

## Part I

```python
# Coordinates
A = np.array([0, 0, 0])
B = np.array([0, 3, 0])
D = np.array([-2, 0, 1.5])
E = np.array([-2, 0, -1.5])
def get_C(theta):
    return 3 * np.array([
        mt.cos(mt.radians(30)) * mt.cos(theta),
        mt.sin(mt.radians(30)),
        mt.cos(mt.radians(30)) * mt.sin(theta)
    ])

# Position vectors
r_AD = D - A
r_AE = E - A
r_DE = E - D
r_BD = D - B
r_BE = E - B
def get_r_AC(theta): return get_C(theta) - A
def get_r_BC(theta): return get_C(theta) - B

# Unit vectors
u_AD = r_AD / np.linalg.norm(r_AD)
u_AE = r_AE / np.linalg.norm(r_AE)
u_DE = r_DE / np.linalg.norm(r_DE)
u_BD = r_BD / np.linalg.norm(r_BD)
u_BE = r_BE / np.linalg.norm(r_BE)
```

```python
# Moment vectors [kN-m] of the load about a point
def M_A(m, theta):
    return m * spc.g * 3 * mt.cos(mt.radians(30)) * np.array([
        mt.sin(theta), 0, -mt.cos(theta)
    ])

def M_B(m, theta): return M_A(m, theta)

def M_E(m, theta):
    return m * spc.g * np.array([
        (3 * mt.cos(mt.radians(30)) * mt.sin(theta)) + 1.5,
        0,
        -(3 * mt.cos(mt.radians(30)) * mt.cos(theta)) - 2
    ])

def M_D(m, theta):
    return -m * spc.g * np.array([
        -((3 * mt.cos(mt.radians(30)) * mt.sin(theta)) + 1.5),
            0,
        -(3 * mt.cos(mt.radians(30)) * mt.cos(theta)) - 2
    ])
```

```python
# Moment vectors [kN-m] of the load about an axis
def M_AD(m, theta): return np.dot(M_A(m, theta), u_AD) * u_AD

def M_AE(m, theta): return np.dot(M_A(m, theta), u_AE) * u_AE

def M_DE(m, theta):
    return np.array([0, 0, -m * spc.g * ((3 * mt.cos(mt.radians(30)) * mt.cos(theta)) + 2)])
```

```python
# Force vectors [kN] acting on (support reactions) and within the derrick
def get_derrick_forces(m, theta):
    r_AC = get_r_AC(theta); r_BC = get_r_BC(theta)
    u_AC = r_AC / np.linalg.norm(r_AC)
    u_BC = r_BC / np.linalg.norm(r_BC)
    w = m * spc.g

    # Solve for tensions in BD and BE from moment-balance about A
    coef = np.array([
        [u_BD[2], u_BE[2]],
        [u_BD[0], u_BE[0]]
    ])
    rhs = np.array([
        -w * mt.cos(mt.radians(30)) * mt.sin(theta),
        -w * mt.cos(mt.radians(30)) * mt.cos(theta)
    ])
    BDBE = np.linalg.solve(coef, rhs)
    T_BD = BDBE[0] * u_BD; T_BE = BDBE[1] * u_BE
    
    # Solve for reaction at A from balance of (external) forces on the derrick
    R_A = -(T_BD + T_BE + (w * np.array([0, -1, 0])))

    # Solve for BC tension and AC force by force balance at C
    T_BC = w * u_BC; F_AC = w * u_AC

    # Solve for AB force by force balance at B
    F_AB = -(T_BD + T_BE + T_BC)

    return T_BD, T_BE, R_A, T_BC, F_AC, F_AB
```

### Scenario 1

```python
m = 1; theta = mt.pi / 6

print(f"Question 1. <{'; '.join(map(str, M_A(m ,theta)))}>")
print(f"Question 2. <{'; '.join(map(str, M_B(m ,theta)))}>")
print(f"Question 3. <{'; '.join(map(str, M_E(m ,theta)))}>")
print(f"Question 4. <{'; '.join(map(str, M_D(m ,theta)))}>")
print(f"Question 5. <{'; '.join(map(str, M_AD(m ,theta)))}>")
print(f"Question 6. <{'; '.join(map(str, M_AE(m ,theta)))}>")
print(f"Question 7. <{'; '.join(map(str, M_DE(m ,theta)))}>")
```

### Scenario 2

```python
m = 1; theta = mt.pi / 3

T_BD, T_BE, R_A, T_BC, F_AC, F_AB = get_derrick_forces(m, theta)

print(f"Question 1. {np.linalg.norm(T_BD)}")
print(f"Question 2. {np.linalg.norm(T_BE)}")
print(f"Question 3. {np.linalg.norm(R_A)}")
print(f"Question 4. {np.linalg.norm(T_BC)}")
print(f"Question 5. {np.linalg.norm(F_AC)}")
print(f"Question 6. {np.linalg.norm(F_AB)}")
```
