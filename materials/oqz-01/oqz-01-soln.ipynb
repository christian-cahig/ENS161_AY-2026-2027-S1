```python
import math as mt

import numpy as np
import scipy.constants as spc
import scipy.optimize as spo
```

## Part I

```python
def T(m_b, m_c, L):
    return m_c * spc.g / 2

def A_x(m_b, m_c, L):
    return mt.sqrt(3) * (m_b - (2*m_c)) * spc.g / 2

def O_x(m_b, m_c, L):
    return mt.sqrt(3) * (m_b - (1.5*m_c)) * spc.g / 2

def O_y(m_b, m_c, L):
    return (m_b - (0.25 * m_c)) * spc.g
```

### Scenario 01

```python
m_b, m_c, L = 4, 1, 1

print(f"Question 1. {T(m_b, m_c, L):.7f}")
print(f"Question 3. <{A_x(m_b, m_c, L):.7f}; {0:.7f}>")
print(f"Question 4. <{-O_x(m_b, m_c, L):.7f}; {O_y(m_b, m_c, L):.7f}>")
```

### Scenario 02

```python
# This is basically doing a shift-solve,
# but I do not have my ClassWiz,
# moreso the patience to do algebraic manipulations.
fun = lambda x : A_x(m_b, x, L)
m_c_max, info = spo.newton(fun, x0 = m_c, full_output=True, disp=False)
print(info)

print(f"Question 1. {m_c_max:.7f}")
print(f"Question 2. {T(m_b, m_c_max, L):.7f}")
print(f"Question 3. <{A_x(m_b, m_c_max, L):.7f}; {0:.7f}>")
print(f"Question 4. <{-O_x(m_b, m_c_max, L):.7f}; {O_y(m_b, m_c_max, L):.7f}>")
```

### Scenario 03

```python
# This is the maximum allowable m_c,
# based solely on impending rotation about O.
fun = lambda x : A_x(m_b, x, L)
m_c_max_rot, info = spo.newton(fun, x0 = m_c, full_output=True, disp=False)
print(info)

# This is the maximum allowable m_c,
# based solely on cable limits.
fun = lambda x : 7 - T(m_b, x, L)
m_c_max_cab, info = spo.newton(fun, x0 = m_c, full_output=True, disp=False)
print(info)

# The maximum allowable m_c is the lesser of the "partial" maximum allowables.
m_c_max = min(m_c_max_rot, m_c_max_cab)

print(f"Question 1. {m_c_max:.7f}")
print(f"Question 2. {T(m_b, m_c_max, L):.7f}")
print(f"Question 3. <{A_x(m_b, m_c_max, L):.7f}; {0:.7f}>")
print(f"Question 4. <{-O_x(m_b, m_c_max, L):.7f}; {O_y(m_b, m_c_max, L):.7f}>")
```
